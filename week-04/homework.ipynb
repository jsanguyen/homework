**Review Problem 1**

**Concept:** *Functions with lists and indexing.*

**Task:**

Write a function middlest_element(values) that returns the middle element of a list.

Rules:

* If the list is empty, return `None`.

* If the list length is odd, return the element at the exact middle index.

* If the list length is even, return the element just left of center (example: for length 6, use index 2).

* Then run the provided test cases and print results exactly as shown.

**Sample Output:**

```
middlest_element([1,2,3,4,5]) =  3
middlest_element([1,2,3,4,5,6]) =  3
middlest_element([]) =  None
```


````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def middlest_element(values):
    """
    Return the middle element of a list.

    Parameters:
        values (list): the list

    Returns:
        The middle element of the list or None if the list is empty.
        - Odd length: returns the element at the exact middle index.
        - Even length: returns the element just left of center
          (for example, length 6 uses index 2).
    """
    # Decision: an empty list has no middle, so return None right away
    if len(values) == 0:
        return None

    # Formula: (length - 1) // 2 finds the middle index for both cases
    #   Odd length 5:  (5 - 1) // 2 = 2  -> exact middle
    #   Even length 6: (6 - 1) // 2 = 2  -> just left of center
    # len() gives the number of elements and // is integer division
    middle_index = (len(values) - 1) // 2

    # Use indexing to get and return the element at the middle index
    return values[middle_index]

````

````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def square_evens(values):
    """
    Return a new list with the square of every even number in values.

    Parameters:
        values (list of numbers): any sequence of numbers

    Returns:
        list: the squares of the even numbers in the same order they
        appear in values. The input is not changed.
    """
    # List comprehension: loop through each element n in values.
    # Decision: keep n only if it is even (n % 2 == 0 means dividing by 2 leaves no remainder).
    # Formula: n ** 2 squares each even number that is kept.
    return [n ** 2 for n in values if n % 2 == 0]
````

````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def print_powers(base, n):
    """
    Print the first n+1 powers of base from exponent 0 up to exponent n.

    Parameters: base (int): the number to be raised to successive powers n (int): the highest exponent to compute (inclusive)

    Returns:
        None 
    """
    # Print the header line; n + 1 is the count of powers because we include exponent 0
    print(f"printing the first {n + 1} powers of {base}")

    # For loop: iterate over every exponent i from 0 up to and including n.
    # range(n + 1) produces 0, 1, 2, ..., n (the stop value is exclusive so we add 1 to reach n itself).
    for i in range(n + 1):
        # Formula: base ** i raises base to the i-th power
        result = base ** i

        print(f"{base} ** {i} = {result}")

print_powers(2, 10)
````

````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def print_names(names):
    """
    Print each name in the list on its own line using an index-based loop.

    Parameters: names (list of str): the names to print

    Returns: None
    """
    # Index-based for loop: i takes on 0, 1, ..., len(names) - 1
    for i in range(len(names)):
        # Use the index to look up the name at position i
        print(names[i])


def print_adjacent_names(names):
    """
    Print each name alongside the next name in the list

    Parameters names (list of str): the names to pair up

    Returns None 
    """
    # Index-based for loop: stop at len(names) - 1 so that names[i + 1]
    for i in range(len(names) - 1):
        # Print the current name and the one right after it
        print(f"{names[i]} , {names[i + 1]}")


# Test code
names = ['John', 'Paul', 'George', 'Ringo', 'Billy']

# Announce the call, then run print_names
print(f"calling print_names({names})")
print_names(names)

# Announce the call, then run print_adjacent_names
print(f"calling print_adjacent_names({names})")
print_adjacent_names(names)
````

Write a function `mean(values)` that returns the arithmetic mean of a list of numbers.

````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def mean(values):
    """
    Return the arithmetic mean of a list of numbers.

    Parameters values (list of int or float): the numbers to average

    Returns float: the sum of the values divided by how many there are
    """
    # Accumulator: start the running total at 0
    total = 0

    # Element-based for loop: add each value to the running total
    for value in values:
        total += value

    # Formula: mean = total / count
    return total / len(values)



````

**Problem 7**

**Concept:** *Accumulator pattern with numeric result, list iteration.*

**Task:**

Write a function `variance(values)` that returns the population variance of a list of numbers using this formula:

$$
\text{variance} = \frac{1}{n} \sum_{i=1}^{n} (x_i - \bar{x})^2
$$

Rules:

* Must use loops + accumulator (no `sum()`).

* No imports.

* Must call your `mean(values)` function to get the mean.

* Then run the provided test case and print in the same format.

**Sample Output:**

```
values: [6, 3, 7, 9, 5]
mean: 6.0
variance: 4.0
```


````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def variance(values):
    """
    Return the population variance of a list of numbers.

    Parameters values (list of int or float): the numbers to measure spread for

    Returns float: the average of the squared deviations from the mean
    """
    # Function call: reuse mean() from Problem 6 to get the average (x-bar)
    avg = mean(values)

    # Accumulator: running total of squared deviations, starting at 0
    total = 0

    # Element-based for loop: add each squared deviation to the total
    for value in values:
        # Formula: (x_i - x-bar) ** 2
        total += (value - avg) ** 2

    # Formula: variance = (1 / n) * sum of squared deviations
    return total / len(values)
````

**Problem 8**

**Concept:** *Accumulator pattern with numeric result, list iteration, index-based loop.*

**Task:**

Write a function `covariance(xs, ys)` that computes the population covariance between two series:

$$
\text{covariance} = \frac{1}{n} \sum_{i=1}^{n} (x_i - \bar{x})(y_i - \bar{y})
$$

Rules:

* If `xs` and `ys` are not the same length, print:
    
    `xs and ys must be the same length`
    
    and return `0`.

* Must use an index-based loop.

* Must use your `mean()` function.

* Then run the given test case and print exactly in the same format.

**Sample Output:**

```
x values: [6, 3, 7, 9, 5]
y values: [4, 1, 7, 5, 0]
covariance: 3.8
```


````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def covariance(xs, ys):
    """
    Return the population covariance between two series of numbers.

    Parameters:
        xs (list of int or float): the first series
        ys (list of int or float): the second series (same length as xs)

    Returns:
        float: the average of the products of paired deviations from the means,
               or 0 if the two lists are not the same length
    """
    # Decision:  lengths must match
    if len(xs) != len(ys):
        print("xs and ys must be the same length")
        return 0

    # Function calls: reuse mean() from Problem 6 to get x-bar and y-bar
    x_avg = mean(xs)
    y_avg = mean(ys)

    # Accumulator: running total of deviation products, starting at 0
    total = 0

    # Index-based for loop: i lets us grab the paired values xs[i] and ys[i]
    for i in range(len(xs)):
        # Formula: (x_i - x-bar) * (y_i - y-bar)
        total += (xs[i] - x_avg) * (ys[i] - y_avg)

    # Formula: covariance = (1 / n) * sum of deviation products
    return total / len(xs)



````

````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def polynomial_value(x, coefficients):
    """
    Parameters:
        x (int or float): the value to plug into the polynomial
        coefficients (list of int or float): coefficients[i] multiplies x**i

    Returns:
        int or float: the value of the polynomial at x
    """
    # Accumulator: running total of the terms starting at 0
    result = 0

    # Index-based for loop: i is both the position in the list AND the exponent
    for i in range(len(coefficients)):
        # Formula: add the term coefficients[i] * x**i
        result += coefficients[i] * x ** i

    return result


````

Write a function `string_to_ascii(s)` that returns a list of ASCII codes for each character in `s`.

Rules:

* Must build a list using an accumulator (ex: `append`).

**Sample Output:**



````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def string_to_ascii(s):
    """
    Return a list of the ASCII codes for each character in a string.

    Parameters:
        s (str): the string to convert

    Returns:
        list of int: the ASCII code of each character, in order
    """
    # Accumulator: start with an empty list to collect the codes
    codes = []

    # Element-based for loop: visit each character in the string
    for ch in s:
        # ord() gives the ASCII code of a single character; append it to the list
        codes.append(ord(ch))

    return codes


````

Write a function `ascii_to_string(codes)` that converts a list of ASCII codes into the corresponding string.

Rules:

**Sample Output:**

````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def ascii_to_string(codes):
    """
    Convert a list of ASCII codes into the corresponding string.

    Parameters:
        codes (list of int): the ASCII codes to convert

    Returns:
        str: the characters for each code joined together in order
    """
    # Accumulator: start with an empty string to build the result
    result = ""

    # Element-based for loop: visit each ASCII code in the list
    for code in codes:
        # chr() is the inverse of ord(): it turns a code into a one-character
        # string; concatenate it onto the end of the result
        result += chr(code)

    return result


````

**Problem 12**

**Concept:** *Accumulator pattern to process a string, produce a string result.*

**Task:**

Write a function `cipher(s, n)` that applies a Caesar cipher shift to a plaintext string.

Rules:

* Assume lowercase letters only for shifting (`a` to `z`).

* Do not shift non-letters (spaces stay spaces, punctuation stays punctuation).

* Must wrap around the alphabet if shifting past `z` or before `a`.

* Must use an accumulator string.

* Then run the exact test cases shown and print in the same format.

**Sample Output:**

```
plaintext = abc, ciphered = def
plaintext = abc xyz, ciphered = def abc
```


````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def cipher(s, n):
    """
    Apply a Caesar cipher shift of n positions to a plaintext string.

    Parameters:
        s (str): the plaintext (lowercase letters, spaces, punctuation)
        n (int): how many positions to shift each letter

    Returns:
        str: the ciphered string, with non-letters left unchanged
    """
    # Accumulator: start with an empty string to build the ciphertext
    result = ""

    # Element-based for loop: visit each character in the plaintext
    for ch in s:
        # Decision: only shift lowercase letters a-z
        if "a" <= ch <= "z":
            # Step 1: convert the letter to a position 0-25 (a=0, b=1, ..., z=25)
            position = ord(ch) - ord("a")
            # Step 2: shift, then wrap with % 26 so 25 + 3 -> 2 (z -> c)
            #         and negative shifts like 0 - 1 -> 25 (a -> z)
            shifted = (position + n) % 26
            # Step 3: convert the position back to a letter and add it on
            result += chr(shifted + ord("a"))
        else:
            # Spaces, punctuation, etc. are copied through unchanged
            result += ch

    return result
````

Write a function `print_all_decipherings(s)` that prints all 26 possible Caesar shifts of string `s`.

Rules:

* For each shift `i`, compute `ciphered = cipher(s, i)` and print:

````python
# TO DO: WRITE YOUR CODE/SOLUTION BELOW
def print_all_decipherings(s):
    """
    Print all 26 possible Caesar shifts of a string one per line.

    Parameters:
        s (str): the (possibly ciphered) text to try every shift on

    Returns:
        None
    """
    # Definite loop: try every shift 0 through 25 (range stops before 26)
    for i in range(26):
        # Function call: reuse cipher() from Problem 12 to apply shift i
        ciphered = cipher(s, i)
        # Print in the required format: shift number, comma, shifted text
        print(f"{i}, {ciphered}")
````
